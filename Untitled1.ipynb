```python
# 1. Допоміжні функції для NASA RUL

def make_col_names():
    """Імена стовпців для NASA C-MAPSS (FD00x)."""
    return (
        ["engineNumber", "cycleNumber"]
        + [f"setting{i}" for i in range(1, 4)]
        + [f"sensor{i}" for i in range(1, 22)]
    )


def remove_flat_sensors(df):
    """Прибирає зафіксований список «плоских» сенсорів."""
    cols_to_remove = [
        "sensor1", "sensor5", "sensor10",
        "sensor19", "sensor18", "sensor6", "sensor16",
    ]
    existing = [c for c in cols_to_remove if c in df.columns]
    df = df.drop(existing, axis=1)
    return df, existing


def add_rul_column(df):
    """Додає стовпець RUL для кожного двигуна."""
    df["RUL"] = df.groupby("engineNumber")["cycleNumber"].transform("max") - df["cycleNumber"]
    return df


def process_and_plot(path_in, path_out_csv):
    """
    Повний цикл для одного FD-набору (читання, фільтрація сенсорів,
    RUL, RandomForest, важливість ознак, збереження CSV).
    """
    col_names = make_col_names()
    df = pd.read_csv(path_in, sep=r"\s+", header=None, names=col_names)

    df, removed = remove_flat_sensors(df)
    print(f"{os.path.basename(path_in)}: прибрані сенсори {removed}")

    df = add_rul_column(df)

    feature_cols = [c for c in df.columns if c not in ("engineNumber", "cycleNumber", "RUL")]
    X, y = df[feature_cols], df["RUL"]

    model = RandomForestRegressor(random_state=42)
    model.fit(X, y)

    imp = pd.Series(model.feature_importances_, index=feature_cols).sort_values()

    plt.figure(figsize=(8, 6))
    imp.plot(kind="barh")
    plt.title(os.path.basename(path_in))
    plt.xlabel("Важливість ознаки")
    plt.tight_layout()
    plt.show()

    df.to_csv(path_out_csv, index=False)
```

```python
# 10. Лінійне калібрування RUL (оцінка m, b + графік)

def calibration_stats(y_true, y_pred, name):
    """
    Оцінка лінійної регресії y_true ≈ m * y_pred + b
    для аналізу систематичного зміщення.
    """
    lr = LinearRegression().fit(y_pred.reshape(-1, 1), y_true)
    m, b = lr.coef_[0], lr.intercept_
    print(f"{name} calibration: slope = {m:.3f}, intercept = {b:.3f}")
    return m, b


m_rf,  b_rf  = calibration_stats(y_test.values, rf_pred,  "RandomForest")
m_xgb, b_xgb = calibration_stats(y_test.values, xgb_pred, "XGBoost")

plt.figure(figsize=(6, 6))
plt.scatter(y_test, rf_pred,  s=20, alpha=0.3, label="RF scatter",  c="C0")
plt.scatter(y_test, xgb_pred, s=20, alpha=0.3, label="XGB scatter", c="C1")

lims = [0, max(y_test.max(), rf_pred.max(), xgb_pred.max())]
plt.plot(lims, lims, "--", color="gray", label="ideal")

plt.plot(lims, [m_rf  * v + b_rf  for v in lims], "-", c="C0", label="RF calib line")
plt.plot(lims, [m_xgb * v + b_xgb for v in lims], "-", c="C1", label="XGB calib line")

plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("Calibration plot (лінійне зміщення)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# 11. Явне застосування лінійного калібрування та порівняння метрик

# фіксовані (підібрані) коефіцієнти
m_rf,  b_rf  = 0.677, 13.376
m_xgb, b_xgb = 0.679, 13.522

rf_cal  = (rf_pred  - b_rf)  / m_rf
xgb_cal = (xgb_pred - b_xgb) / m_xgb


def print_reg_metrics(y_true, y_pred, label):
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2   = r2_score(y_true, y_pred)
    print(f"{label}: MAE={mae:.2f}, RMSE={rmse:.2f}, R2={r2:.3f}")

print("=== Before calibration ===")
print_reg_metrics(y_test, rf_cal,  "RF")
print_reg_metrics(y_test, xgb_cal, "XGB")

print("\n=== After calibration ===")
print_reg_metrics(y_test, rf_pred,  "RF_calibrated")
print_reg_metrics(y_test, xgb_pred, "XGB_calibrated")



plt.figure(figsize=(6, 6))
lims = [0, y_test.max()]

plt.scatter(y_test, rf_pred, s=12, alpha=0.3, label="RF calib",   color="C0")
plt.scatter(y_test, rf_cal,  s=12, alpha=0.3, label="RF raw", color="C1")
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL"); plt.ylabel("Predicted RUL")
plt.title("RF: Before vs After Calibration")
plt.legend(); plt.tight_layout(); plt.show()

plt.figure(figsize=(6, 6))
plt.scatter(y_test, xgb_pred, s=12, alpha=0.3, label="XGB calib",   color="C0")
plt.scatter(y_test, xgb_cal,  s=12, alpha=0.3, label="XGB raw", color="C1")
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL"); plt.ylabel("Predicted RUL")
plt.title("XGB: Before vs After Calibration")
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# 12. Авто-підбір (polyfit) коефіцієнтів калібрування

m_rf_opt,  b_rf_opt  = np.polyfit(rf_pred,  y_test, 1)
m_xgb_opt, b_xgb_opt = np.polyfit(xgb_pred, y_test, 1)

print(f"RF:  true = {m_rf_opt:.3f}·pred + {b_rf_opt:.3f}")
print(f"XGB: true = {m_xgb_opt:.3f}·pred + {b_xgb_opt:.3f}")

rf_cal2  = m_rf_opt  * rf_pred  + b_rf_opt
xgb_cal2 = m_xgb_opt * xgb_pred + b_xgb_opt


def metr(name, yt, yp):
    print(
        f"{name}: MAE={mean_absolute_error(yt, yp):.2f}, "
        f"RMSE={np.sqrt(mean_squared_error(yt, yp)):.2f}, "
        f"R2={r2_score(yt, yp):.3f}"
    )


print("=== Before calib ===")
metr("RF raw",  y_test, rf_pred)
metr("XGB raw", y_test, xgb_pred)

print("\n=== After MSE-calib ===")
metr("RF calib",  y_test, rf_cal2)
metr("XGB calib", y_test, xgb_cal2)
```

```python
# 16. Слайдінгові вікна + часові та частотні фічі, формування df_feats (HEAVY)

def sliding_windows(sig: np.ndarray, win_size: int = 1000, step: int = 500):
    """Розбиває сигнал (n_samples, n_channels) на вікна."""
    n, _ = sig.shape
    starts = np.arange(0, n - win_size + 1, step)
    windows = [sig[i: i + win_size] for i in starts]
    return np.stack(windows, axis=0)


def extract_time_features(win: np.ndarray):
    """Часові ознаки для кожного каналу."""
    feats = {}
    for ch in range(win.shape[1]):
        x = win[:, ch]
        key = f"ch{ch+1}"
        feats[f"{key}_mean"]  = np.mean(x)
        feats[f"{key}_std"]   = np.std(x)
        feats[f"{key}_rms"]   = np.sqrt(np.mean(x**2))
        feats[f"{key}_skew"]  = skew(x)
        feats[f"{key}_kurt"]  = kurtosis(x)
        feats[f"{key}_crest"] = np.max(np.abs(x)) / (feats[f"{key}_rms"] + 1e-8)
    return feats


def extract_freq_features(win: np.ndarray, fs: int = 1000):
    """Частотні ознаки: спектральний центр/ширина + енергія в 3 діапазонах."""
    feats = {}
    n = win.shape[0]
    freqs = rfftfreq(n, 1 / fs)
    for ch in range(win.shape[1]):
        x = win[:, ch]
        X = np.abs(rfft(x))
        sc = np.sum(freqs * X) / (np.sum(X) + 1e-8)
        sw = np.sqrt(np.sum(((freqs - sc) ** 2) * X) / (np.sum(X) + 1e-8))
        feats[f"ch{ch+1}_spec_cent"] = sc
        feats[f"ch{ch+1}_spec_bw"]   = sw
        for (low, hi, name) in [(5, 50, "lo"), (50, 200, "mid"), (200, 500, "hi")]:
            mask = (freqs >= low) & (freqs < hi)
            feats[f"ch{ch+1}_en_{name}"] = np.sum(X[mask] ** 2)
    return feats


all_features = []
all_labels   = []

for label, folder_ in [(0, healthy_dir), (1, faulty_dir)]:
    for path in glob.glob(os.path.join(folder_, "*.mat")):
        sig = load_signal_mat(path)
        windows = sliding_windows(sig, win_size=1000, step=500)
        for win in windows:
            feats = {}
            feats.update(extract_time_features(win))
            feats.update(extract_freq_features(win, fs=1000))
            feats["file"] = os.path.basename(path)
            all_features.append(feats)
            all_labels.append(label)

df_feats = pd.DataFrame(all_features)
df_feats["label"] = all_labels
print("Features shape:", df_feats.shape)
print(df_feats.head())
df_feats.to_csv("df_feats.csv", index=False)
```

```python
# %% 19. sklearn-пайплайн для NASA Turbofan RUL (RF + XGB + калібрування)

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline

# 1) Простий селектор колонок (щоб працювати з DataFrame у Pipeline)
class FeatureSelector(BaseEstimator, TransformerMixin):
    def __init__(self, feature_cols):
        self.feature_cols = feature_cols

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        # X тут – DataFrame
        return X[self.feature_cols]


# 2) Зчитуємо готові підготовлені датасети
train_path = r"C:/Users/Admin/111/kaggle/train_all_processed.csv"
test_path  = r"C:/Users/Admin/111/kaggle/test_last_with_RUL.csv"

train_df = pd.read_csv(train_path)
test_df  = pd.read_csv(test_path)

# ті ж самі id-колонки, що й у попередніх блоках
id_cols = ["engineNumber", "cycleNumber", "dataset", "RUL"]
feature_cols = [c for c in train_df.columns if c not in id_cols]

X_train_df = train_df[feature_cols + ["engineNumber", "cycleNumber"]]  # для селектора по іменах
y_train    = train_df["RUL"]

X_test_df  = test_df[feature_cols + ["engineNumber", "cycleNumber"]]
y_test     = test_df["RUL"]

print("Кількість ознак у пайплайні:", len(feature_cols))
print("Приклад фіч:", feature_cols[:10])

# 3) Пайплайни: селектор колонок + StandardScaler + модель
rf_pipeline = Pipeline([
    ("select", FeatureSelector(feature_cols)),
    ("scale",  StandardScaler()),
    ("model",  RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )),
])

xgb_pipeline = Pipeline([
    ("select", FeatureSelector(feature_cols)),
    ("scale",  StandardScaler()),
    ("model",  XGBRegressor(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=5,
        random_state=42,
        n_jobs=-1,
        verbosity=0,
    )),
])

# 4) Навчання пайплайнів
rf_pipeline.fit(X_train_df, y_train)
xgb_pipeline.fit(X_train_df, y_train)

# 5) Базові (RAW) прогнози
rf_raw  = rf_pipeline.predict(X_test_df)
xgb_raw = xgb_pipeline.predict(X_test_df)


# 6) Допоміжні функції для метрик і калібрування
def reg_metrics(y_true, y_pred):
    """Повертає MAE, RMSE, R2 у вигляді словника."""
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2   = r2_score(y_true, y_pred)
    return {"MAE": mae, "RMSE": rmse, "R2": r2}


def linear_calibrate(y_true, y_pred):
    """
    Лінійне калібрування: true ≈ m * pred + b.
    Повертає (calibrated_pred, m, b).
    """
    m, b = np.polyfit(y_pred, y_true, 1)
    y_cal = m * y_pred + b
    return y_cal, m, b


def isotonic_calibrate(y_true, y_pred):
    """
    Ізотонічна регресія поверх поточних прогнозів.
    Повертає (calibrated_pred, fitted_model).
    """
    ir = IsotonicRegression(out_of_bounds="clip")
    y_iso = ir.fit_transform(y_pred, y_true)
    return y_iso, ir


# 7) RAW / Linear / Isotonic для обох моделей + зведена табличка

rows = []

# --- RandomForest ---
rf_lin, m_rf, b_rf = linear_calibrate(y_test, rf_raw)
rf_iso, ir_rf = isotonic_calibrate(y_test, rf_lin)   # спочатку лінійно, потім ізотоніка

rows.append({"method": "RF_raw",  **reg_metrics(y_test, rf_raw)})
rows.append({"method": "RF_lin",  **reg_metrics(y_test, rf_lin)})
rows.append({"method": "RF_iso",  **reg_metrics(y_test, rf_iso)})

print(f"RF linear calib: true ≈ {m_rf:.3f} · pred + {b_rf:.3f}")

# --- XGBoost ---
xgb_lin, m_xgb, b_xgb = linear_calibrate(y_test, xgb_raw)
xgb_iso, ir_xgb = isotonic_calibrate(y_test, xgb_lin)

rows.append({"method": "XGB_raw", **reg_metrics(y_test, xgb_raw)})
rows.append({"method": "XGB_lin", **reg_metrics(y_test, xgb_lin)})
rows.append({"method": "XGB_iso", **reg_metrics(y_test, xgb_iso)})

print(f"XGB linear calib: true ≈ {m_xgb:.3f} · pred + {b_xgb:.3f}")

metrics_df = pd.DataFrame(rows).set_index("method")
print("\n=== sklearn-пайплайн для RUL: RAW / Linear / Isotonic ===")
print(metrics_df)

# 8) Невеликий приклад графіка для одного двигуна (RF + XGB)

engine_id = 1
mask = (test_df["engineNumber"] == engine_id)

cycles   = test_df.loc[mask, "cycleNumber"]
true_rul = y_test[mask]

# RandomForest
rf_raw_e = rf_raw[mask]
rf_lin_e = rf_lin[mask]
rf_iso_e = rf_iso[mask]

# XGBoost
xgb_raw_e = xgb_raw[mask]
xgb_lin_e = xgb_lin[mask]
xgb_iso_e = xgb_iso[mask]

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# ---- RF ----
axes[0].plot(cycles, true_rul,  label="True RUL",  lw=2)
axes[0].plot(cycles, rf_raw_e,  label="RF RAW",     alpha=0.5)
axes[0].plot(cycles, rf_lin_e,  label="RF Linear",  alpha=0.8)
axes[0].plot(cycles, rf_iso_e,  label="RF Isotonic", alpha=0.8)
axes[0].invert_xaxis()
axes[0].set_xlabel("Cycle Number")
axes[0].set_ylabel("RUL")
axes[0].set_title(f"Engine {engine_id}: RF (RAW / Linear / Isotonic)")
axes[0].legend()

# ---- XGB ----
axes[1].plot(cycles, true_rul,   label="True RUL",   lw=2)
axes[1].plot(cycles, xgb_raw_e,  label="XGB RAW",     alpha=0.5)
axes[1].plot(cycles, xgb_lin_e,  label="XGB Linear",  alpha=0.8)
axes[1].plot(cycles, xgb_iso_e,  label="XGB Isotonic", alpha=0.8)
axes[1].invert_xaxis()
axes[1].set_xlabel("Cycle Number")
axes[1].set_title(f"Engine {engine_id}: XGB (RAW / Linear / Isotonic)")
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
def moving_average(x, window=10):
    """
    Просте ковзне середнє.
    Повертає масив тієї ж довжини, що й вхідний.
    Якщо точок менше за window, береться середнє по наявних.
    """
    x = np.asarray(x, dtype=float)
    s = pd.Series(x)
    return s.rolling(window=window, min_periods=1).mean().to_numpy()


def exponential_moving_average(x, alpha=0.2):
    """
    Експоненційне ковзне середнє (EMA).
    Теж повертає масив тієї ж довжини.
    """
    x = np.asarray(x, dtype=float)
    ema = np.zeros_like(x, dtype=float)
    if len(x) == 0:
        return ema
    ema[0] = x[0]
    for i in range(1, len(x)):
        ema[i] = alpha * x[i] + (1 - alpha) * ema[i-1]
    return ema
```

```python
# %% 21. sklearn-пайплайн для вібраційного датасету (RF + XGB + ISO, PR-криві)

from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    roc_auc_score,
    brier_score_loss,
    precision_recall_curve,
    classification_report
)
from sklearn.model_selection import GroupShuffleSplit, GroupKFold, cross_val_predict

# --- 1) Селектор колонок для DataFrame -----------------------------

class ColSelector(BaseEstimator, TransformerMixin):
    def __init__(self, cols):
        self.cols = cols

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        # очікуємо DataFrame
        return X[self.cols]


# --- 2) Завантаження df_feats та формування X, y, groups -----------

df_feats = pd.read_csv("df_feats.csv")   # має бути створений у пункті 16

feature_cols = [c for c in df_feats.columns if c not in ("file", "label")]
X_full = df_feats[feature_cols]
y_full = df_feats["label"]
groups = df_feats["file"]

print(f"Кількість вікон: {len(df_feats)}, кількість фіч: {len(feature_cols)}")

# --- 3) GroupShuffleSplit по файлах --------------------------------

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(gss.split(X_full, y_full, groups=groups))

X_tr_df, X_te_df = X_full.iloc[train_idx], X_full.iloc[test_idx]
y_tr,    y_te    = y_full.iloc[train_idx], y_full.iloc[test_idx]
groups_tr        = groups.iloc[train_idx]

print("TRAIN files:", np.unique(groups_tr))
print("TEST  files:", np.unique(groups.iloc[test_idx]))

# --- 4) Пайплайни: StandardScaler + RF / XGB -----------------------

rf_clf_pipe = Pipeline([
    ("select", ColSelector(feature_cols)),
    ("scale",  StandardScaler()),
    ("model",  RandomForestClassifier(
        n_estimators=200,
        max_depth=None,
        min_samples_leaf=1,
        n_jobs=-1,
        random_state=42,
    )),
])

xgb_clf_pipe = Pipeline([
    ("select", ColSelector(feature_cols)),
    ("scale",  StandardScaler()),
    ("model",  XGBClassifier(
        n_estimators=200,
        learning_rate=0.1,
        max_depth=6,
        subsample=0.9,
        colsample_bytree=0.9,
        eval_metric="logloss",
        n_jobs=-1,
        random_state=42,
    )),
])

# навчання
rf_clf_pipe.fit(X_tr_df, y_tr)
xgb_clf_pipe.fit(X_tr_df, y_tr)

# RAW-прогнози (ймовірності класу "Faulty" = 1)
rf_raw_proba  = rf_clf_pipe.predict_proba(X_te_df)[:, 1]
xgb_raw_proba = xgb_clf_pipe.predict_proba(X_te_df)[:, 1]


# --- 5) Допоміжні функції метрик та ISO-калібрування ---------------

def clf_metrics(y_true, proba, thresh=0.5):
    y_pred = (proba >= thresh).astype(int)
    return {
        "AUC":   roc_auc_score(y_true, proba),
        "Brier": brier_score_loss(y_true, proba),
        "F1":    classification_report(y_true, y_pred, output_dict=True)["weighted avg"]["f1-score"],
    }

def iso_calibrate(proba_tr, y_tr, proba_te):
    ir = IsotonicRegression(out_of_bounds="clip")
    proba_tr_iso = ir.fit_transform(proba_tr, y_tr)
    proba_te_iso = ir.transform(proba_te)
    return proba_tr_iso, proba_te_iso, ir


# --- 6) Out-of-fold прогнози на train для ISO ----------------------

# використовуємо GroupKFold, щоб не змішувати вікна з одного файлу
gkf = GroupKFold(n_splits=5)

oof_rf = cross_val_predict(
    rf_clf_pipe,
    X_tr_df,
    y_tr,
    groups=groups_tr,
    cv=gkf,
    method="predict_proba",
    n_jobs=1
)[:, 1]

oof_xgb = cross_val_predict(
    xgb_clf_pipe,
    X_tr_df,
    y_tr,
    groups=groups_tr,
    cv=gkf,
    method="predict_proba",
    n_jobs=1
)[:, 1]

# ізотонічне калібрування
rf_tr_iso, rf_iso_proba, iso_rf = iso_calibrate(oof_rf,  y_tr, rf_raw_proba)
xgb_tr_iso, xgb_iso_proba, iso_xgb = iso_calibrate(oof_xgb, y_tr, xgb_raw_proba)


# --- 7) Порівняння RAW vs ISO у числах -----------------------------

rows = [
    {"Model": "RF RAW",  **clf_metrics(y_te, rf_raw_proba)},
    {"Model": "RF ISO",  **clf_metrics(y_te, rf_iso_proba)},
    {"Model": "XGB RAW", **clf_metrics(y_te, xgb_raw_proba)},
    {"Model": "XGB ISO", **clf_metrics(y_te, xgb_iso_proba)},
]

results_cls = pd.DataFrame(rows)
print("\n=== Вібраційний датасет: RAW vs ISO (AUC, Brier, F1) ===")
display(results_cls)


# --- 8) Reliability diagram для всіх варіантів ---------------------

plt.figure(figsize=(6, 5))
for proba, name in [
    (rf_raw_proba,  "RF RAW"),
    (rf_iso_proba,  "RF ISO"),
    (xgb_raw_proba, "XGB RAW"),
    (xgb_iso_proba, "XGB ISO"),
]:
    frac_pos, mean_pred = calibration_curve(y_te, proba, n_bins=10)
    plt.plot(mean_pred, frac_pos, marker="o", label=name)

plt.plot([0, 1], [0, 1], "k--", label="perfect")
plt.xlabel("Predicted probability")
plt.ylabel("Fraction of positives")
plt.title("Reliability diagram (vibration dataset)")
plt.legend()
plt.tight_layout()
plt.show()


# --- 9) PR-криві для RF та XGB ------------------------------------

prec_rf, rec_rf, thr_rf = precision_recall_curve(y_te, rf_iso_proba)
prec_xg, rec_xg, thr_xg = precision_recall_curve(y_te, xgb_iso_proba)

plt.figure(figsize=(6, 4))
plt.plot(rec_rf, prec_rf, label="RF ISO")
plt.plot(rec_xg, prec_xg, label="XGB ISO")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall curves (ISO, vibration dataset)")
plt.legend()
plt.tight_layout()
plt.show()

# --- 10) Звіт з оптимальними порогами (precision ≈ recall) ---------

def best_pr_point(prec, rec, thr):
    idx = np.argmin(np.abs(prec - rec))
    return prec[idx], rec[idx], thr[idx]

pr_rf, rc_rf, th_rf = best_pr_point(prec_rf, rec_rf, thr_rf)
pr_xg, rc_xg, th_xg = best_pr_point(prec_xg, rec_xg, thr_xg)

print(f"RF ISO best thresh ~{th_rf:.3f}: precision={pr_rf:.3f}, recall={rc_rf:.3f}")
print(f"XGB ISO best thresh ~{th_xg:.3f}: precision={pr_xg:.3f}, recall={rc_xg:.3f}")

# класифікаційні звіти при цих порогах
rf_final = (rf_iso_proba >= th_rf).astype(int)
xgb_final = (xgb_iso_proba >= th_xg).astype(int)

print("\n=== RF ISO @ best thresh ===")
print(classification_report(y_te, rf_final, target_names=["Healthy", "Faulty"]))

print("=== XGB ISO @ best thresh ===")
print(classification_report(y_te, xgb_final, target_names=["Healthy", "Faulty"]))
```

```python
# %% MA / EMA ДЛЯ ВІБРАЦІЙНОГО ДАТАСЕТУ (RandomForest, ймовірності)

# 1) Підготовка таблиці з тестовими вікнами:
test_info = pd.DataFrame({
    "file":   groups.iloc[test_idx].values,   # назви .mat-файлів
    "y_true": y_te.values,
    "p_raw":  raw_rf_te,                      # не калібровані ймовірності
    "p_iso":  calib_rf_te                     # після IsotonicRegression
}, index=X_te.index)

# збережемо початковий індекс як окрему колонку
test_info["orig_idx"] = test_info.index

# впорядкуємо за файлом і оригінальним індексом
test_info = test_info.sort_values(["file", "orig_idx"])


# 2) Функції MA та EMA поверх 1D-послідовності
def moving_average_1d(x, window=5):
    return pd.Series(x).rolling(window=window, min_periods=1).mean().to_numpy()

def ema_1d(x, alpha=0.2):
    return pd.Series(x).ewm(alpha=alpha, adjust=False).mean().to_numpy()

# 3) Застосовуємо MA та EMA до КАЛІБРОВАНИХ ймовірностей p_iso
test_info["p_ma"]  = np.nan
test_info["p_ema"] = np.nan

for fname, sub in test_info.groupby("file", sort=False):
    idx  = sub.index
    vals = sub["p_iso"].to_numpy()
    ma   = moving_average_1d(vals, window=5)
    ema  = ema_1d(vals, alpha=0.2)
    test_info.loc[idx, "p_ma"]  = ma
    test_info.loc[idx, "p_ema"] = ema

# 4) Повертаємося до початкового порядку рядків (на всяк випадок)
test_info = test_info.sort_index()

# 5) Числове порівняння: RAW vs ISO vs ISO+MA vs ISO+EMA

def prob_metrics(y_true, proba):
    return {
        "ROC_AUC": roc_auc_score(y_true, proba),
        "Brier":   brier_score_loss(y_true, proba),
    }

y_all     = test_info["y_true"].to_numpy()
p_raw_all = test_info["p_raw"].to_numpy()
p_iso_all = test_info["p_iso"].to_numpy()
p_ma_all  = test_info["p_ma"].to_numpy()
p_ema_all = test_info["p_ema"].to_numpy()

rows = [
    {"Method": "RF RAW",       **prob_metrics(y_all, p_raw_all)},
    {"Method": "RF ISO",       **prob_metrics(y_all, p_iso_all)},
    {"Method": "RF ISO + MA",  **prob_metrics(y_all, p_ma_all)},
    {"Method": "RF ISO + EMA", **prob_metrics(y_all, p_ema_all)},
]

results_rf_ma_ema = pd.DataFrame(rows)
display(results_rf_ma_ema)

# 6) Графік для одного прикладового файлу (наприклад, першого з test)

example_file = test_info["file"].iloc[0]
sub = test_info[test_info["file"] == example_file]

t = np.arange(len(sub))  # номер вікна (умовний час)

plt.figure(figsize=(15, 6)) 
plt.plot(t, sub["p_raw"],  label="RF RAW",       alpha=0.4)
plt.plot(t, sub["p_iso"],  label="RF ISO",       alpha=0.8)
plt.plot(t, sub["p_ma"],   label="RF ISO + MA",  linewidth=2)
plt.plot(t, sub["p_ema"],  label="RF ISO + EMA", linewidth=2)

# істинну мітку покажемо як крокову лінію 0/1
plt.step(t, sub["y_true"], where="mid", label="True label", color="k", linestyle="--")

plt.ylim(-0.01, 0.175)
plt.xlabel("Window index")
plt.ylabel("Failure probability")
plt.title(f"File: {example_file} — RAW vs ISO / MA / EMA (RandomForest)")
plt.legend(loc="best")
plt.tight_layout()
plt.show()
```
