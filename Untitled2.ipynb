```python
# ==============================================================================
# КОМІРКА 1: ІМПОРТИ ТА КОНФІГУРАЦІЯ
# ==============================================================================

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pywt # НОВИЙ ІМПОРТ: Для вейвлет-деноізингу
import pickle # Для збереження моделей

# SciPy
from scipy.io import loadmat
from scipy.signal import butter, filtfilt
from scipy.stats import skew, kurtosis
from scipy.fft import rfft, rfftfreq

# Scikit-learn
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.isotonic import IsotonicRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    roc_auc_score,
    classification_report,
    brier_score_loss,
    precision_recall_curve,
)
from sklearn.model_selection import (
    train_test_split,
    GroupShuffleSplit,
    RandomizedSearchCV,
    cross_val_predict,
)
from sklearn.calibration import calibration_curve
from sklearn.linear_model import LinearRegression

# XGBoost
from xgboost import XGBRegressor, XGBClassifier


# --- ГЛОБАЛЬНІ КОНСТАНТИ ---
DATA_ROOT = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
DATA_PATH_RUL = os.path.join(DATA_ROOT, 'kaggle/')
DATA_PATH_VIBRATION = os.path.join(DATA_ROOT, 'kaggle2/')
FS_VIB = 1000 # Частота дискретизації вібраційних даних (Гц)
```

```python
# ==============================================================================
# КОМІРКА 2: ДОПОМІЖНІ ФУНКЦІЇ (Утиліти та Feature Engineering)
# ==============================================================================

def make_col_names():
    """Формує імена стовпців для NASA C-MAPSS."""
    return (
        ["engineNumber", "cycleNumber"]
        + [f"setting{i}" for i in range(1, 4)]
        + [f"sensor{i}" for i in range(1, 22)]
    )


def remove_flat_sensors(df):
    """Прибирає зафіксований список «плоских» сенсорів."""
    cols_to_remove = [
        "sensor1", "sensor5", "sensor10", "sensor19", "sensor18", "sensor6", "sensor16",
    ]
    existing = [c for c in cols_to_remove if c in df.columns]
    df = df.drop(existing, axis=1)
    return df, existing


def add_rul_column(df):
    """Додає стовпець RUL для кожного двигуна."""
    df["RUL"] = df.groupby("engineNumber")["cycleNumber"].transform("max") - df["cycleNumber"]
    return df


def generate_advanced_rolling_features(df, window_size=20):
    """
    ПОКРАЩЕННЯ A: Генерує розширені ковзні ознаки (mean, std, skew, kurtosis)
    для RUL-прогнозу.
    """
    df_out = df.copy()
    feature_cols = [c for c in df.columns if c.startswith('sensor') or c.startswith('setting')]
    
    # ФІКС: Примусове перетворення в число
    for col in feature_cols:
        df_out[col] = pd.to_numeric(df_out[col], errors='coerce')
    df_out.dropna(subset=feature_cols, inplace=True) 

    groups = df_out.groupby('engineNumber')
    
    for col in feature_cols:
        roll = groups[col].transform(lambda x: x.rolling(window=window_size, min_periods=1))
        
        df_out[f'roll_mean_{col}'] = roll.mean()
        df_out[f'roll_std_{col}'] = roll.std().fillna(0)
        df_out[f'roll_skew_{col}'] = roll.apply(lambda x: x.skew() if len(x) > 2 else 0, raw=True).fillna(0)
        df_out[f'roll_kurt_{col}'] = roll.apply(lambda x: x.kurtosis() if len(x) > 3 else 0, raw=True).fillna(0)

    # print(f"\n[INFO] Додано {len(feature_cols) * 4} розширених ковзних ознак з вікном {window_size}.")
    return df_out


def load_signal_mat(filepath):
    """Завантажує .mat-файл і повертає сигнали (X/Y/Z)."""
    mat = loadmat(filepath)
    key = next(k for k in mat.keys() if not k.startswith("__"))
    return mat[key][:, :3]


def bandpass(sig, low=5, high=100, fs=FS_VIB, order=4):
    """Band-pass фільтр Баттерворта для одного каналу (СТАНДАРТ)."""
    nyq = 0.5 * fs
    b, a = butter(order, [low / nyq, high / nyq], btype="band")
    return filtfilt(b, a, sig)


def wavelet_denoise(sig, wavelet='db4', level=4, threshold_factor=0.5):
    """
    ПОКРАЩЕННЯ C: Простий денойзинг за допомогою дискретного вейвлет-перетворення (DWT).
    """
    coeffs = pywt.wavedec(sig, wavelet, level=level)
    sigma = np.median(np.abs(coeffs[-1])) / 0.6745
    threshold = sigma * threshold_factor * np.sqrt(2 * np.log(len(sig)))
    coeffs[1:] = (pywt.threshold(c, threshold, mode='soft') for c in coeffs[1:])
    return pywt.waverec(coeffs, wavelet)


def process_and_plot(path_in, path_out_csv):
    """
    Повний цикл для одного FD-набору: читання, обробка, тренування RF,
    малювання важливості ознак та збереження обробленого CSV.
    """
    col_names = make_col_names()
    
    # ФІКС: Примусове обрізання DataFrame до 26 стовпців для уникнення TypeError
    df = pd.read_csv(path_in, sep=r"\s+", header=None, engine='python')
    df = df.iloc[:, 0:26]
    df.columns = col_names 
    df.dropna(axis=1, how='all', inplace=True) 
    df.columns = df.columns.str.strip() 

    df, removed = remove_flat_sensors(df)
    
    # --- ВИКЛИК ПОКРАЩЕННЯ A ---
    df = generate_advanced_rolling_features(df, window_size=20)
    
    print(f"\n{os.path.basename(path_in)}: прибрані сенсори {removed}")

    df = add_rul_column(df)

    feature_cols = [c for c in df.columns if c not in ("engineNumber", "cycleNumber", "RUL")]
    X, y = df[feature_cols], df["RUL"]

    # --- ТРЕНУВАННЯ ТА ВИВІД ВАЖЛИВОСТІ ОЗНАК ---
    model = RandomForestRegressor(random_state=42)
    model.fit(X, y)

    imp = pd.Series(model.feature_importances_, index=feature_cols).sort_values()

    plt.figure(figsize=(8, 6))
    imp.tail(30).plot(kind="barh") 
    plt.title(f"Важливість ознак для {os.path.basename(path_in)} (з розширеними ковзними)")
    plt.xlabel("Важливість ознаки")
    plt.ylabel("Ознака")
    plt.tight_layout()
    plt.show()

    df.to_csv(path_out_csv, index=False)


def extract_time_features(win: np.ndarray):
    """Часові ознаки для кожного каналу."""
    feats = {}
    for ch in range(win.shape[1]):
        x = win[:, ch]
        key = f"ch{ch+1}"
        feats[f"{key}_mean"] = np.mean(x)
        feats[f"{key}_std"] = np.std(x)
        feats[f"{key}_rms"] = np.sqrt(np.mean(x**2))
        feats[f"{key}_skew"] = skew(x)
        feats[f"{key}_kurt"] = kurtosis(x)
        feats[f"{key}_crest"] = np.max(np.abs(x)) / (feats[f"{key}_rms"] + 1e-8)
    return feats


def extract_advanced_freq_features(win: np.ndarray, fs: int = FS_VIB):
    """
    ПОКРАЩЕННЯ B: Розширені частотні ознаки (енергія в смугах).
    Викликається в КОМІРЦІ 16.
    """
    feats = {}
    n = win.shape[0]
    freqs = rfftfreq(n, 1 / fs)
    
    BAND_DEFECT = (100, 300) 
    BAND_DISBALANCE = (0, 50) 

    for ch in range(win.shape[1]):
        x = win[:, ch]
        X = np.abs(rfft(x))
        key = f"ch{ch+1}"
        
        sc = np.sum(freqs * X) / (np.sum(X) + 1e-8)
        sw = np.sqrt(np.sum(((freqs - sc) ** 2) * X) / (np.sum(X) + 1e-8))
        feats[f"{key}_spec_cent"] = sc
        feats[f"{key}_spec_bw"] = sw
        
        mask_defect = (freqs >= BAND_DEFECT[0]) & (freqs < BAND_DEFECT[1])
        feats[f"{key}_en_defect"] = np.sum(X[mask_defect] ** 2)
        
        mask_low = (freqs >= BAND_DISBALANCE[0]) & (freqs < BAND_DISBALANCE[1])
        feats[f"{key}_en_low"] = np.sum(X[mask_low] ** 2)
        
        feats[f"{key}_en_ratio"] = feats[f"{key}_en_defect"] / (feats[f"{key}_en_low"] + 1e-8)
        
    return feats


def calibration_stats(y_true, y_pred, name):
    """Оцінка лінійної регресії y_true = m * y_pred + b для аналізу зміщення."""
    lr = LinearRegression().fit(y_pred.reshape(-1, 1), y_true)
    m, b = lr.coef_[0], lr.intercept_
    print(f"{name} calibration: slope = {m:.3f}, intercept = {b:.3f}")
    return m, b

def print_reg_metrics(y_true, y_pred, label):
    """Виводить метрики MAE, RMSE, R2 для регресії."""
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"{label}: MAE={mae:.2f}, RMSE={rmse:.2f}, R2: {r2:.3f}")

def report_split(name, y_true, y_pred):
    """Виводить метрики для порівняння train/test (Секція 12)."""
    print(
        f"{name} — MAE: {mean_absolute_error(y_true, y_pred):.2f}, "
        f"RMSE: {np.sqrt(mean_squared_error(y_true, y_pred)):.2f}, "
        f"R2: {r2_score(y_true, y_pred):.3f}"
    )

def metr(name, yt, yp):
    """Виводить метрики для калібрування (Секція 10)."""
    print(
        f"{name}: MAE={mean_absolute_error(yt, yp):.2f}, "
        f"RMSE={np.sqrt(mean_squared_error(yt, yp)):.2f}, "
        f"R2={r2_score(yt, yp):.3f}"
    )

def print_iso_metrics(prefix, y_true, y_raw, y_iso):
    """Виводить метрики для ізотонічного калібрування RUL (Секція 11)."""
    print(f"--- {prefix} ---")
    print(
        " raw: MAE={:.2f}, RMSE={:.2f}, R2={:.3f}".format(
            mean_absolute_error(y_true, y_raw),
            np.sqrt(mean_squared_error(y_true, y_raw)),
            r2_score(y_true, y_raw),
        )
    )
    print(
        " iso: MAE={:.2f}, RMSE={:.2f}, R2={:.3f}\n".format(
            mean_absolute_error(y_true, y_iso),
            np.sqrt(mean_squared_error(y_true, y_iso)),
            r2_score(y_true, y_iso),
        )
    )

def sliding_windows(sig: np.ndarray, win_size: int = 1000, step: int = 500):
    """Розбиває сигнал (n_samples, n_channels) на вікна."""
    n, _ = sig.shape
    starts = np.arange(0, n - win_size + 1, step)
    windows = [sig[i : i + win_size] for i in starts]
    return np.stack(windows, axis=0)
```

```python
# ==============================================================================
# КОМІРКА 1: ІМПОРТИ ТА КОНФІГУРАЦІЯ
# ==============================================================================

import os
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pywt 
import pickle 

# SciPy
from scipy.io import loadmat
from scipy.signal import butter, filtfilt, savgol_filter 
from scipy.stats import skew, kurtosis
from scipy.fft import rfft, rfftfreq

# Scikit-learn
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.isotonic import IsotonicRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    roc_auc_score,
    classification_report,
    brier_score_loss,
    precision_recall_curve,
)
from sklearn.model_selection import (
    train_test_split,
    GroupShuffleSplit,
    RandomizedSearchCV,
    cross_val_predict,
)
from sklearn.calibration import calibration_curve
from sklearn.linear_model import LinearRegression

# XGBoost
from xgboost import XGBRegressor, XGBClassifier


# --- ГЛОБАЛЬНІ КОНСТАНТИ ---
DATA_ROOT = os.path.dirname(os.path.abspath(__file__)) if '__file__' in locals() else os.getcwd()
DATA_PATH_RUL = os.path.join(DATA_ROOT, 'kaggle/')
DATA_PATH_VIBRATION = os.path.join(DATA_ROOT, 'kaggle2/')
FS_VIB = 1000 # Частота дискретизації вібраційних даних (Гц)
RUL_WINDOW = 20 # Розмір вікна для ковзних статистик


# ==============================================================================
# КОМІРКА 2: ДОПОМІЖНІ ФУНКЦІЇ (Утиліти та Feature Engineering)
# ==============================================================================

def make_col_names():
    """Формує імена стовпців для NASA C-MAPSS."""
    return (
        ["engineNumber", "cycleNumber"]
        + [f"setting{i}" for i in range(1, 4)]
        + [f"sensor{i}" for i in range(1, 22)]
    )


def remove_flat_sensors(df):
    """Прибирає зафіксований список «плоских» сенсорів."""
    cols_to_remove = [
        "sensor1", "sensor5", "sensor10", "sensor19", "sensor18", "sensor6", "sensor16",
    ]
    existing = [c for c in cols_to_remove if c in df.columns]
    df = df.drop(existing, axis=1)
    return df, existing


def add_rul_column(df):
    """Додає стовпець RUL для кожного двигуна."""
    df["RUL"] = df.groupby("engineNumber")["cycleNumber"].transform("max") - df["cycleNumber"]
    return df


def generate_advanced_rolling_features(df, window_size=RUL_WINDOW):
    """
    ПОКРАЩЕННЯ A: Генерує розширені ковзні ознаки (mean, std, skew, kurtosis)
    для RUL-прогнозу.
    """
    df_out = df.copy()
    feature_cols = [c for c in df.columns if c.startswith('sensor') or c.startswith('setting')]
    
    # ПРИПУСКАЄМО, ЩО ПЕРЕТВОРЕННЯ В pd.to_numeric ВІДБУЛОСЯ ДО ЦЬОГО ВИКЛИКУ.
    
    groups = df_out.groupby('engineNumber')
    
    for col in feature_cols:
        # Увага: roll тепер працює з числовим стовпцем
        roll = groups[col].transform(lambda x: x.rolling(window=window_size, min_periods=1))
        
        # 1. EMA (Експоненційне Ковзне Середнє)
        df_out[f'ema_mean_{col}'] = groups[col].transform(
            lambda x: x.ewm(span=window_size, min_periods=1).mean()
        )
        # 2. Ковзне стандартне відхилення
        df_out[f'roll_std_{col}'] = roll.std().fillna(0)
        
        # 3. Ковзний перекіс (Skewness)
        df_out[f'roll_skew_{col}'] = roll.apply(lambda x: x.skew() if len(x) > 2 else 0, raw=True).fillna(0)
        
        # 4. Ковзний ексцес (Kurtosis)
        df_out[f'roll_kurt_{col}'] = roll.apply(lambda x: x.kurtosis() if len(x) > 3 else 0, raw=True).fillna(0)
        
        # 5. Savitzky–Golay Filter (для згладжування)
        df_out[f'sg_filt_{col}'] = groups[col].transform(
            lambda x: savgol_filter(x.values.astype(float), window_length=11, polyorder=3) if len(x) >= 11 else x
        )
        

    return df_out


def generate_ema_comparison_features(df, method='MA', window_size=RUL_WINDOW):
    """
    НОВА ФУНКЦІЯ (КОМІРКА 21): Генерує ознаки RUL з використанням або MA, або EMA
    для прямого порівняння ефектів згладжування.
    """
    df_out = df.copy()
    feature_cols = [c for c in df.columns if c.startswith('sensor') or c.startswith('setting')]
    
    # ФІКС: ПРИМУСОВЕ ЧИЩЕННЯ
    for col in feature_cols:
        df_out[col] = pd.to_numeric(df_out[col], errors='coerce', downcast='float')
    df_out.dropna(subset=feature_cols, inplace=True) 

    groups = df_out.groupby('engineNumber')
    
    for col in feature_cols:
        # 1. СТАТИСТИКИ (STD, SKEW, KURT) залишаємо спільними
        roll = groups[col].transform(lambda x: x.rolling(window=window_size, min_periods=1))
        df_out[f'roll_std_{col}'] = roll.std().fillna(0)
        df_out[f'roll_skew_{col}'] = roll.apply(lambda x: x.skew() if len(x) > 2 else 0, raw=True).fillna(0)
        df_out[f'roll_kurt_{col}'] = roll.apply(lambda x: x.kurtosis() if len(x) > 3 else 0, raw=True).fillna(0)
        
        # 2. ЗГЛАДЖЕННЯ (ЗМІНЮЄТЬСЯ)
        if method == 'MA':
            # Просте ковзне середнє
            df_out[f'mean_{col}'] = roll.mean()
        elif method == 'EMA':
            # Експоненційне ковзне середнє (більша вага новим даним)
            df_out[f'mean_{col}'] = groups[col].transform(
                lambda x: x.ewm(span=window_size, min_periods=1).mean()
            )
        
    return df_out


def load_signal_mat(filepath):
    """Завантажує .mat-файл і повертає сигнали (X/Y/Z)."""
    mat = loadmat(filepath)
    key = next(k for k in mat.keys() if not k.startswith("__"))
    return mat[key][:, :3]


def wavelet_denoise(sig, wavelet='db4', level=4, threshold_factor=0.5):
    """Простий денойзинг за допомогою дискретного вейвлет-перетворення (DWT)."""
    coeffs = pywt.wavedec(sig, wavelet, level=level)
    sigma = np.median(np.abs(coeffs[-1])) / 0.6745
    threshold = sigma * threshold_factor * np.sqrt(2 * np.log(len(sig)))
    coeffs[1:] = (pywt.threshold(c, threshold, mode='soft') for c in coeffs[1:])
    return pywt.waverec(coeffs, wavelet)


def process_and_plot(path_in, path_out_csv):
    """
    Повний цикл для одного FD-набору: читання, обробка, тренування RF,
    малювання важливості ознак та збереження обробленого CSV.
    """
    col_names = make_col_names()
    
    # 1. ЧИТАННЯ ТА БАЗОВЕ ЧИЩЕННЯ
    df = pd.read_csv(path_in, sep=r"\s+", header=None, engine='python')
    df = df.iloc[:, 0:26]
    df.columns = col_names 
    df.dropna(axis=1, how='all', inplace=True) 
    df.columns = df.columns.str.strip() 
    
    # 2. ВИДАЛЕННЯ ПЛОСКИХ СЕНСОРІВ
    df, removed = remove_flat_sensors(df)
    
    # 3. АГРЕСИВНЕ ПЕРЕТВОРЕННЯ ТИПІВ (ФІНАЛЬНИЙ ФІКС)
    feature_cols = [c for c in df.columns if c.startswith('sensor') or c.startswith('setting')]
    for col in feature_cols:
        # Примусове перетворення в float, щоб уникнути TypeError в rolling
        df[col] = pd.to_numeric(df[col], errors='coerce', downcast='float')
    df.dropna(subset=feature_cols, inplace=True)

    # --- ВИКЛИК ПОКРАЩЕННЯ A ---
    df = generate_advanced_rolling_features(df, window_size=RUL_WINDOW)
    
    print(f"\n{os.path.basename(path_in)}: прибрані сенсори {removed}")

    df = add_rul_column(df)

    feature_cols = [c for c in df.columns if c not in ("engineNumber", "cycleNumber", "RUL")]
    X, y = df[feature_cols], df["RUL"]

    # --- ТРЕНУВАННЯ ТА ВИВІД ВАЖЛИВОСТІ ОЗНАК ---
    model = RandomForestRegressor(random_state=42)
    model.fit(X, y)

    imp = pd.Series(model.feature_importances_, index=feature_cols).sort_values()

    plt.figure(figsize=(8, 6))
    imp.tail(30).plot(kind="barh") 
    plt.title(f"Важливість ознак для {os.path.basename(path_in)} (з розширеними ковзними)")
    plt.xlabel("Важливість ознаки")
    plt.ylabel("Ознака")
    plt.tight_layout()
    plt.show()

    df.to_csv(path_out_csv, index=False)


def calibration_stats(y_true, y_pred, name):
    """Оцінка лінійної регресії y_true = m * y_pred + b для аналізу зміщення."""
    lr = LinearRegression().fit(y_pred.reshape(-1, 1), y_true)
    m, b = lr.coef_[0], lr.intercept_
    print(f"{name} calibration: slope = {m:.3f}, intercept = {b:.3f}")
    return m, b

def print_reg_metrics(y_true, y_pred, label):
    """Виводить метрики MAE, RMSE, R2 для регресії."""
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"{label}: MAE={mae:.2f}, RMSE={rmse:.2f}, R2: {r2:.3f}")

def report_split(name, y_true, y_pred):
    """Виводить метрики для порівняння train/test (Секція 12)."""
    print(
        f"{name} — MAE: {mean_absolute_error(y_true, y_pred):.2f}, "
        f"RMSE: {np.sqrt(mean_squared_error(y_true, y_pred)):.2f}, "
        f"R2: {r2_score(y_true, y_pred):.3f}"
    )

def metr(name, yt, yp):
    """Виводить метрики для калібрування (Секція 10)."""
    print(
        f"{name}: MAE={mean_absolute_error(yt, yp):.2f}, "
        f"RMSE={np.sqrt(mean_squared_error(yt, yp)):.2f}, "
        f"R2={r2_score(yt, yp):.3f}"
    )

def print_iso_metrics(prefix, y_true, y_raw, y_iso):
    """Виводить метрики для ізотонічного калібрування RUL (Секція 11)."""
    print(f"--- {prefix} ---")
    print(
        " raw: MAE={:.2f}, RMSE={:.2f}, R2={:.3f}".format(
            mean_absolute_error(y_true, y_raw),
            np.sqrt(mean_squared_error(y_true, y_raw)),
            r2_score(y_true, y_raw),
        )
    )
    print(
        " iso: MAE={:.2f}, RMSE={:.2f}, R2={:.3f}\n".format(
            mean_absolute_error(y_true, y_iso),
            np.sqrt(mean_squared_error(y_true, y_iso)),
            r2_score(y_true, y_iso),
        )
    )

def sliding_windows(sig: np.ndarray, win_size: int = 1000, step: int = 500):
    """Розбиває сигнал (n_samples, n_channels) на вікна."""
    n, _ = sig.shape
    starts = np.arange(0, n - win_size + 1, step)
    windows = [sig[i : i + win_size] for i in starts]
    return np.stack(windows, axis=0)


def extract_time_features(win: np.ndarray):
    """Часові ознаки для кожного каналу."""
    feats = {}
    for ch in range(win.shape[1]):
        x = win[:, ch]
        key = f"ch{ch+1}"
        feats[f"{key}_mean"] = np.mean(x)
        feats[f"{key}_std"] = np.std(x)
        feats[f"{key}_rms"] = np.sqrt(np.mean(x**2))
        feats[f"{key}_skew"] = skew(x)
        feats[f"{key}_kurt"] = kurtosis(x)
        feats[f"{key}_crest"] = np.max(np.abs(x)) / (feats[f"{key}_rms"] + 1e-8)
    return feats


def extract_advanced_freq_features(win: np.ndarray, fs: int = FS_VIB):
    """
    ПОКРАЩЕННЯ B: Розширені частотні ознаки (енергія в смугах).
    """
    feats = {}
    n = win.shape[0]
    freqs = rfftfreq(n, 1 / fs)
    
    BAND_DEFECT = (100, 300) 
    BAND_DISBALANCE = (0, 50) 

    for ch in range(win.shape[1]):
        x = win[:, ch]
        X = np.abs(rfft(x))
        key = f"ch{ch+1}"
        
        sc = np.sum(freqs * X) / (np.sum(X) + 1e-8)
        sw = np.sqrt(np.sum(((freqs - sc) ** 2) * X) / (np.sum(X) + 1e-8))
        feats[f"{key}_spec_cent"] = sc
        feats[f"{key}_spec_bw"] = sw
        
        mask_defect = (freqs >= BAND_DEFECT[0]) & (freqs < BAND_DEFECT[1])
        feats[f"{key}_en_defect"] = np.sum(X[mask_defect] ** 2)
        
        mask_low = (freqs >= BAND_DISBALANCE[0]) & (freqs < BAND_DISBALANCE[1])
        feats[f"{key}_en_low"] = np.sum(X[mask_low] ** 2)
        
        feats[f"{key}_en_ratio"] = feats[f"{key}_en_defect"] / (feats[f"{key}_en_low"] + 1e-8)
        
    return feats


# ==============================================================================
# КОМІРКА 3: ПОПЕРЕДНЯ ОБРОБКА ОДНОГО TEST_FD001 + ВАЖЛИВІСТЬ ОЗНАК (Секція 1)
# ==============================================================================

print("--- Секція 3: Демонстраційна обробка та важливість ознак FD001 ---")

# Зчитуємо test_FD001
path_test_fd001 = os.path.join(DATA_PATH_RUL, "test_FD001.txt")
col_names = make_col_names()
df_test_single = pd.read_csv(
    path_test_fd001,
    sep=r"\s+",
    header=None,
    names=col_names,
    engine='python'
)

# --- ВИВІД 1 ---
print("\n[Вивід 1: Перші рядки сирих даних test_FD001]")
print(df_test_single.head().to_markdown(index=False))

df_test_single, removed = remove_flat_sensors(df_test_single)

# 3. АГРЕСИВНЕ ПЕРЕТВОРЕННЯ ТИПІВ (ФІНАЛЬНИЙ ФІКС)
feature_cols_all = [c for c in df_test_single.columns if c.startswith('sensor') or c.startswith('setting')]
for col in feature_cols_all:
    df_test_single[col] = pd.to_numeric(df_test_single[col], errors='coerce', downcast='float')
df_test_single.dropna(subset=feature_cols_all, inplace=True)


# --- ВИКЛИК ПОКРАЩЕННЯ A ---
df_test_single = generate_advanced_rolling_features(df_test_single, window_size=RUL_WINDOW)

print("\nПрибрані стовпці (малоінформативні):", removed)
print("Залишились стовпці:", df_test_single.columns.tolist())

# Створення RUL (тільки для демонстрації важливості ознак)
df_test_single = add_rul_column(df_test_single)

feature_columns = [col for col in df_test_single.columns if col not in ["engineNumber", "cycleNumber", "RUL"]]
X_train_single = df_test_single[feature_columns]
y_train_single = df_test_single["RUL"]

# --- ТРЕНУВАННЯ RANDOM FOREST ДЛЯ ВИЗНАЧЕННЯ ВАЖЛИВОСТІ ---
model_rf_imp = RandomForestRegressor(random_state=42)
model_rf_imp.fit(X_train_single, y_train_single)

importances = model_rf_imp.feature_importances_
feature_importance = pd.Series(importances, index=feature_columns)

# --- ВИВІД 2: Важливість ознак ---
plt.figure(figsize=(10, 8))
feature_importance.sort_values().tail(30).plot(kind="barh") 
plt.title("Важливість ознак для прогнозу RUL (RandomForest) на test_FD001 (з розширеними ковзними)")
plt.xlabel("Важливість")
plt.ylabel("Ознака")
plt.tight_layout()
plt.show()

# Зберігаємо проміжний файл для подальшого використання (як у оригіналі)
df_test_single.to_csv(os.path.join(DATA_PATH_RUL, "test_1_demo_with_roll.csv"), index=False)
```

```python
# ==============================================================================
# КОМІРКА 9: ТРЕНУВАННЯ МОДЕЛЕЙ RUL (RF та XGB) та БАЗОВІ МЕТРИКИ (Секція 7)
# ==============================================================================

print("\n--- Секція 9: Тренування та оцінка RUL-моделей ---")

# Завантаження фінальних наборів (з розширеними ковзними ознаками)
train = pd.read_csv(os.path.join(DATA_PATH_RUL, "train_all_processed_roll.csv"))
test = pd.read_csv(os.path.join(DATA_PATH_RUL, "test_last_with_RUL.csv"))

# --- ВИКЛИК ПОКРАЩЕННЯ A (ДЛЯ TEST) ---
test = generate_advanced_rolling_features(test, window_size=20) 
test.drop(columns=['dataset'], errors='ignore', inplace=True) # Видаляємо зайвий стовпець

feature_cols = [
    c
    for c in train.columns
    if c not in ["engineNumber", "cycleNumber", "dataset", "RUL"]
]

# Масштабування X_test за параметрами train
scaler_final = StandardScaler()
train[feature_cols] = scaler_final.fit_transform(train[feature_cols])
test[feature_cols] = scaler_final.transform(test[feature_cols])

X_train = train[feature_cols]
y_train = train["RUL"]
X_test = test[feature_cols]
y_test = test["RUL"]

# 1. RandomForestRegressor
rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print("\n=== RandomForest (RAW) ===")
rf_mae_raw = mean_absolute_error(y_test, rf_pred)
rf_rmse_raw = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2_raw = r2_score(y_test, rf_pred)
print_reg_metrics(y_test, rf_pred, "RF RAW")

# 2. XGBRegressor
xgb = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42,
    n_jobs=-1,
    verbosity=0,
    objective='reg:squarederror'
)
xgb.fit(X_train, y_train)
xgb_pred = xgb.predict(X_test)

print("\n=== XGBoost (RAW) ===")
xgb_mae_raw = mean_absolute_error(y_test, xgb_pred)
xgb_rmse_raw = np.sqrt(mean_squared_error(y_test, xgb_pred))
xgb_r2_raw = r2_score(y_test, xgb_pred)
print_reg_metrics(y_test, xgb_pred, "XGB RAW")

# --- ВИВІД 6 ---
results = pd.DataFrame(
    {"y_true": y_test, "RF_pred": rf_pred, "XGB_pred": xgb_pred}
).reset_index(drop=True)
print("\n[Вивід 6: Пара прикладів прогнозів (True vs Pred)]")
print(results.head(10).to_markdown(index=False))

# --- ГРАФІК True vs Predicted RUL ---
plt.figure(figsize=(8, 8))
plt.scatter(y_test, rf_pred, s=20, alpha=0.5, label="Random Forest")
plt.scatter(y_test, xgb_pred, s=20, alpha=0.5, label="XGBoost")

min_val = min(y_test.min(), rf_pred.min(), xgb_pred.min())
max_val = max(y_test.max(), rf_pred.max(), xgb_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], linestyle="--", color='k')

plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("True vs Predicted Remaining Useful Life (RAW)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# КОМІРКА 10: ЛІНІЙНЕ КАЛІБРУВАННЯ RUL-МОДЕЛЕЙ (Секція 8)
# ==============================================================================

print("\n--- Секція 10: Аналіз систематичного зміщення (Лінійна регресія) ---")

# Оцінка параметрів m та b для корекції зміщення
m_rf, b_rf = calibration_stats(y_test.values, rf_pred, "RandomForest")
m_xgb, b_xgb = calibration_stats(y_test.values, xgb_pred, "XGBoost")

# --- ГРАФІК Лінійного калібрування ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, rf_pred, s=20, alpha=0.3, label="RF scatter", c="C0")
plt.scatter(y_test, xgb_pred, s=20, alpha=0.3, label="XGB scatter", c="C1")
lims = [
    0,
    max(y_test.max(), rf_pred.max(), xgb_pred.max()),
]
plt.plot(lims, lims, "--", color="gray", label="Ideal line (y=x)")
plt.plot(lims, [m_rf * lim + b_rf for lim in lims], "-", c="C0", label="RF calib line")
plt.plot(
    lims, [m_xgb * lim + b_xgb for lim in lims], "-", c="C1", label="XGB calib line"
)
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("Лінійна оцінка калібрування (Y_true vs Y_pred)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# КОМІРКА 11: ЗАСТОСУВАННЯ ЛІНІЙНОГО КАЛІБРУВАННЯ (Секція 9)
# ==============================================================================

print("\n--- Секція 11: Явне застосування лінійного калібрування ---")

# Використовуємо оцінені коефіцієнти (з КОМІРКИ 10) для калібрування прогнозів
rf_cal = m_rf * rf_pred + b_rf
xgb_cal = m_xgb * xgb_pred + b_xgb


print("\n=== Before calibration ===")
print_reg_metrics(y_test, rf_pred, "RF")
print_reg_metrics(y_test, xgb_pred, "XGB")

print("\n=== After Linear Calibration ===")
print_reg_metrics(y_test, rf_cal, "RF_calibrated")
print_reg_metrics(y_cal, xgb_cal, "XGB_calibrated")

# --- ГРАФІК RF: Raw vs Calib ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, rf_pred, s=12, alpha=0.3, label="RF raw", color="C0")
plt.scatter(y_test, rf_cal, s=12, alpha=0.3, label="RF calib (Linear)", color="C1")
lims = [0, y_test.max()]
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("RF: Before vs After Linear Calibration")
plt.legend()
plt.tight_layout()
plt.show()

# --- ГРАФІК XGB: Raw vs Calib ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, xgb_pred, s=12, alpha=0.3, label="XGB raw", color="C0")
plt.scatter(y_test, xgb_cal, s=12, alpha=0.3, label="XGB calib (Linear)", color="C1")
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("XGB: Before vs After Linear Calibration")
plt.legend()
plt.tight_layout()
plt.show()


# ==============================================================================
# КОМІРКА 12: АВТО-ПІДБІР КОЕФІЦІЄНТІВ ЧЕРЕЗ POLYFIT (Секція 10)
# ==============================================================================

print("\n--- Секція 12: Автоматичний підбір коефіцієнтів (Polyfit) ---")

m_rf_opt, b_rf_opt = np.polyfit(rf_pred, y_test, 1)
m_xgb_opt, b_xgb_opt = np.polyfit(xgb_pred, y_test, 1)

print(f"RF (Optimal Polyfit): true = {m_rf_opt:.3f}*pred + {b_rf_opt:.3f}")
print(f"XGB (Optimal Polyfit): true = {m_xgb_opt:.3f}*pred + {b_xgb_opt:.3f}")

# Калібрування за оптимальними параметрами
rf_cal_opt = m_rf_opt * rf_pred + b_rf_opt
xgb_cal_opt = m_xgb_opt * xgb_pred + b_xgb_opt

print("\n=== Before calib ===")
metr("RF raw", y_test, rf_pred)
metr("XGB raw", y_test, xgb_pred)

print("\n=== After MSE-calib (Polyfit) ===")
metr("RF calib", y_test, rf_cal_opt)
metr("XGB calib", y_test, xgb_cal_opt)


# ==============================================================================
# КОМІРКА 13: ІЗОТОНІЧНЕ КАЛІБРУВАННЯ RUL-ПРОГНОЗІВ (Секція 11)
# ==============================================================================

print("\n--- Секція 13: Ізотонічне калібрування (Non-linear correction) ---")

ir_rf = IsotonicRegression(out_of_bounds="clip")
ir_xgb = IsotonicRegression(out_of_bounds="clip")

rf_iso = ir_rf.fit_transform(rf_pred, y_test)
xgb_iso = ir_xgb.fit_transform(xgb_pred, y_test)

print_iso_metrics("RandomForest", y_test, rf_pred, rf_iso)
print_iso_metrics("XGBoost", y_test, xgb_pred, xgb_iso)

# --- ГРАФІК RF: Raw vs Isotonic ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, rf_pred, s=12, alpha=0.3, label="RF raw", color="C0")
plt.scatter(y_test, rf_iso, s=12, alpha=0.3, label="RF iso", color="C1")
lims = [0, y_test.max()]
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("RF: Raw vs Isotonic Calibration")
plt.legend()
plt.tight_layout()
plt.show()

# --- ГРАФІК XGB: Raw vs Isotonic ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, xgb_pred, s=12, alpha=0.3, label="XGB raw", color="C0")
plt.scatter(y_test, xgb_iso, s=12, alpha=0.3, label="XGB iso", color="C1")
plt.plot(lims, lims, "--", color="gray")
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("XGB: Raw vs Isotonic Calibration")
plt.legend()
plt.tight_layout()
plt.show()


# ==============================================================================
# КОМІРКА 14: ОЦІНКА OVER/UNDER-FIT (Секція 12)
# ==============================================================================

print("\n--- Секція 14: Оцінка перенавчання (Overfitting) ---")

y_train_pred_rf = rf.predict(X_train)
y_train_pred_xgb = xgb.predict(X_train)


print("=== TRAIN SET ===")
report_split("RF train", y_train, y_train_pred_rf)
report_split("XGB train", y_train, y_train_pred_xgb)

y_test_pred_rf = rf.predict(X_test)
y_test_pred_xgb = xgb.predict(X_test)

print("\n=== TEST SET (raw) ===")
report_split("RF test", y_test, y_test_pred_rf)
report_split("XGB test", y_test, y_test_pred_xgb)

# --- ГРАФІК True vs Pred (TEST) ---
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_test_pred_rf, s=20, alpha=0.4, label="RF raw")
plt.scatter(y_test, y_test_pred_xgb, s=20, alpha=0.4, label="XGB raw")
lims = [0, max(y_test.max(), y_test_pred_rf.max(), y_test_pred_xgb.max())]
plt.plot(lims, lims, "--", color="k")
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("Raw predictions (Test Set)")
plt.legend()
plt.show()

# --- ГІСТОГРАМА ЗАЛИШКІВ RF ---
residuals_rf = y_test - y_test_pred_rf
plt.figure(figsize=(5, 3))
plt.hist(residuals_rf, bins=30, edgecolor="k")
plt.title("RF residuals (test) - Розподіл похибок")
plt.xlabel("True – Pred (Цикли)")
plt.show()


# ==============================================================================
# КОМІРКА 15: РОБОТА З ВІБРАЦІЙНИМ ДАТАСЕТОМ: QC ТА EDA (Секція 13)
# ==============================================================================

print("\n--- Секція 15: Обробка вібраційних даних (Завантаження та QC) ---")

healthy_dir = os.path.join(DATA_PATH_VIBRATION, "Healthy")
faulty_dir = os.path.join(DATA_PATH_VIBRATION, "Faulty")

healthy_files = glob.glob(os.path.join(healthy_dir, "*.mat"))
faulty_files = glob.glob(os.path.join(faulty_dir, "*.mat"))
print(f"Знайдено {len(healthy_files)} healthy файлів, {len(faulty_files)} faulty файлів.")

# QC та вивід форми файлів
print("\n=== Healthy (QC перевірка форми) ===")
for f in healthy_files[:5]:
    try:
        mat = loadmat(f)
        key = next(k for k in mat.keys() if not k.startswith("__"))
        signal = mat[key]
        print(f"{os.path.basename(f)} → shape = {signal.shape}")
    except Exception:
        print(f"{os.path.basename(f)} → ERROR loading")

print("\n=== Faulty (QC перевірка форми) ===")
for f in faulty_files[:5]:
    try:
        mat = loadmat(f)
        key = next(k for k in mat.keys() if not k.startswith("__"))
        signal = mat[key]
        print(f"{os.path.basename(f)} → shape = {signal.shape}")
    except Exception:
        print(f"{os.path.basename(f)} → ERROR loading")


# --- ГРАФІКИ Сигналів (Healthy vs Faulty) ---
fig, axs = plt.subplots(3, 2, figsize=(12, 8), sharex=True)
plt.suptitle("Демонстрація сирих вібраційних сигналів (X, Y, Z)")
for i, path in enumerate(healthy_files[:3]):
    sig = load_signal_mat(path)
    axs[i, 0].plot(sig)
    axs[i, 0].set_title(f"Healthy: {os.path.basename(path)}")
for i, path in enumerate(faulty_files[:3]):
    sig = load_signal_mat(path)
    axs[i, 1].plot(sig)
    axs[i, 1].set_title(f"Faulty: {os.path.basename(path)}")
plt.tight_layout()
plt.show()

# --- ГІСТОГРАМА та BOXPLOT (EDA) ---
dfs_v = []
# Обмежуємо кількість файлів для EDA для швидкодії
for label, paths in [("Healthy", healthy_files[:30]), ("Faulty", faulty_files[:30])]:
    for p in paths:
        sig = load_signal_mat(p)
        df_tmp = pd.DataFrame(sig, columns=["X", "Y", "Z"])
        df_tmp["Label"] = label
        dfs_v.append(df_tmp)
df_vals = pd.concat(dfs_v, ignore_index=True)

# Гістограма
g = sns.FacetGrid(df_vals, col="Label", height=4, aspect=1)
g.map_dataframe(lambda data, color: sns.histplot(data["X"], kde=False, bins=50), color="C0")
g.set_axis_labels("Amplitude (X-axis)", "Count")
plt.show()

# Boxplot
plt.figure(figsize=(6, 4))
sns.boxplot(x="Label", y="X", data=df_vals)
plt.title("Amplitude Distribution (X-axis) by Label")
plt.show()


# ==============================================================================
# КОМІРКА 16: СЛАЙДІНГОВІ ВІКНА + ВИЛУЧЕННЯ ФІЧ (Секція 14)
# ==============================================================================

print("\n--- Секція 16: Вилучення ознак із слайдінгових вікон ---")

all_features = []
all_labels = []

for label, folder_ in [(0, healthy_dir), (1, faulty_dir)]:
    for path in glob.glob(os.path.join(folder_, "*.mat")):
        sig = load_signal_mat(path)
        windows = sliding_windows(sig, win_size=1000, step=500)
        for win in windows:
            feats = {}
            feats.update(extract_time_features(win))
            # --- ЗАМІНА: Використовуємо розширені частотні ознаки ---
            feats.update(extract_advanced_freq_features(win, fs=FS_VIB)) 
            feats["file"] = os.path.basename(path)
            all_features.append(feats)
            all_labels.append(label)

df_feats = pd.DataFrame(all_features)
df_feats["label"] = all_labels

# --- ВИВІД 6 ---
print("Features shape:", df_feats.shape)
print("\n[Вивід 6: Перші 5 рядків вилучених ознак (З розширеними частотними)]")
print(df_feats.head().to_markdown(index=False))

# Зберігаємо проміжний файл
df_feats.to_csv("df_feats.csv", index=False)


# ==============================================================================
# КОМІРКА 17: МАСШТАБУВАННЯ ФІЧ ВІБРАЦІЙ + БАЗОВІ МОДЕЛІ (Секція 15)
# ==============================================================================

print("\n--- Секція 17: Масштабування та базове тренування класифікаторів ---")
df_feats = pd.read_csv("df_feats.csv")
# Увага: feature_cols тепер включає і розширені частотні ознаки
feature_cols = [c for c in df_feats.columns if c not in ("label", "file")]
X_raw = df_feats[feature_cols]
y_raw = df_feats["label"]

# --- ВИВІД 7: Статистика RMS до/після масштабування ---
# Динамічно шукаємо всі стовпці, що містять 'rms' (універсальний пошук для 3-х каналів)
rms_cols = [c for c in X_raw.columns if '_rms' in c] 

if not rms_cols:
    print("\n[ПОПЕРЕДЖЕННЯ] Не знайдено стовпців RMS. Перевірте, чи правильно виконана КОМІРКА 16.")
    X = X_raw 
else:
    # Якщо RMS знайдено, виконуємо стандартний вивід та масштабування
    std_scaler = StandardScaler()
    X_std = std_scaler.fit_transform(X_raw)
    df_std = pd.DataFrame(X_std, columns=X_raw.columns)

    print("Before scaling (RMS statistics):")
    print(X_raw[rms_cols].describe().T.to_markdown()) 

    print("\nAfter StandardScaler (RMS statistics):")
    print(df_std[rms_cols].describe().T.to_markdown())
    
    df = df_std.copy()  
    df["label"] = y_raw.values
    X = df.drop("label", axis=1)

y = y_raw # y завжди залишається y_raw


# Простий train/test split
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 1. Random Forest Classifier
rf_clf = RandomForestClassifier(n_estimators=100, n_jobs=-1, random_state=42)
rf_clf.fit(X_tr, y_tr)
rf_pred = rf_clf.predict(X_te)
rf_proba = rf_clf.predict_proba(X_te)[:, 1]

print("\n=== RandomForest (Базова) ===")
print("ROC AUC:", roc_auc_score(y_te, rf_proba))
print("[Classification Report RF]:")
print(classification_report(y_te, rf_pred, target_names=["Healthy", "Faulty"]))

# 2. XGBoost Classifier
xgb_clf = XGBClassifier(eval_metric="logloss", random_state=42, n_jobs=-1)
xgb_clf.fit(X_tr, y_tr)
xgb_pred = xgb_clf.predict(X_te)
xgb_proba = xgb_clf.predict_proba(X_te)[:, 1]

print("\n=== XGBoost (Базова) ===")
print("ROC AUC:", roc_auc_score(y_te, xgb_proba))
print("[Classification Report XGB]:")
print(classification_report(y_te, xgb_pred, target_names=["Healthy", "Faulty"]))

# --- Гіперпараметричне налаштування (RandomizedSearchCV) ---
param_dist = {
    "n_estimators": [100, 200, 500],
    "max_depth": [None, 5, 10, 20],
    "min_samples_leaf": [1, 2, 5],
    "max_features": ["sqrt", "log2", 0.5],
}
rs = RandomizedSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    param_dist,
    n_iter=20,
    cv=3,
    scoring="roc_auc",
    n_jobs=-1,
    random_state=42,
)
rs.fit(X_tr, y_tr)
print("\nBest RF params (Tuning):", rs.best_params_)
best_rf = rs.best_estimator_
print(
    "Tuned RF ROC AUC:",
    roc_auc_score(y_te, best_rf.predict_proba(X_te)[:, 1]),
)


# ==============================================================================
# КОМІРКА 18: GROUP SHUFFLE SPLIT + КАЛІБРУВАННЯ ЙМОВІРНОСТЕЙ (Секція 16)
# ==============================================================================

print("\n--- Секція 18: Калібрування ймовірностей з Group Shuffle Split ---")

# Забезпечуємо, що файли не потрапляють одночасно в train та test
# ПРИМІТКА: feature_cols та X_raw визначаються вище в КОМІРЦІ 17
df_feats_scaled = pd.DataFrame(X, columns=feature_cols)
df_feats_scaled["label"] = y
df_feats_scaled["file"] = df_feats["file"] # Використовуємо файл як групу

X = df_feats_scaled[feature_cols]
y = df_feats_scaled["label"]
groups = df_feats_scaled["file"] # Групування за файлом

gss = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

X_tr, X_te = X.iloc[train_idx], X.iloc[test_idx]
y_tr, y_te = y.iloc[train_idx], y.iloc[test_idx]
groups_te = groups.iloc[test_idx]

print(f"Розмір тренувальної вибірки: {X_tr.shape[0]} вікон.")
print(f"Розмір тестової вибірки: {X_te.shape[0]} вікон.")
print("TRAIN files:", np.unique(groups.iloc[train_idx]))
print("TEST files:", np.unique(groups_te))

# Використовуємо найкращий RF з попередньої секції
rf.fit(X_tr, y_tr)
rf_proba = rf.predict_proba(X_te)[:, 1]

# XGBoost
xgb_final = XGBClassifier(eval_metric="logloss", n_jobs=-1, random_state=42)
xgb_final.fit(X_tr, y_tr)
xgb_proba = xgb_final.predict_proba(X_te)[:, 1]


# --- ІЗОТОНІЧНЕ КАЛІБРУВАННЯ ЙМОВІРНОСТЕЙ ---
raw_rf_te = rf_proba
raw_xgb_te = xgb_proba

# OOF (Out-of-Fold) прогнози для тренування калібратора
oof_rf = cross_val_predict(
    rf, X_tr, y_tr, cv=5, method="predict_proba", n_jobs=-1
)[:, 1]
oof_xgb = cross_val_predict(
    xgb_final, X_tr, y_tr, cv=5, method="predict_proba", n_jobs=-1
)[:, 1]

# Тренування калібратора
iso_rf = IsotonicRegression(out_of_bounds="clip").fit(oof_rf, y_tr)
iso_xgb = IsotonicRegression(out_of_bounds="clip").fit(oof_xgb, y_tr)

# Трансформація тестових ймовірностей
calib_rf_te = iso_rf.transform(raw_rf_te)
calib_xgb_te = iso_xgb.transform(raw_xgb_te)


# --- ВИВІД 8 ---
print("\n=== RandomForest (Probability Calibration) ===")
print(f"ROC-AUC raw: {roc_auc_score(y_te, raw_rf_te):.3f}")
print(f"ROC-AUC iso: {roc_auc_score(y_te, calib_rf_te):.3f}")
print(f"Brier score raw (менше - краще): {brier_score_loss(y_te, raw_rf_te):.4f}")
print(f"Brier score iso (менше - краще): {brier_score_loss(y_te, calib_rf_te):.4f}\n")

print("=== XGBoost (Probability Calibration) ===")
print(f"ROC-AUC raw: {roc_auc_score(y_te, raw_xgb_te):.3f}")
print(f"ROC-AUC iso: {roc_auc_score(y_te, calib_xgb_te):.3f}")
print(f"Brier score raw (менше - краще): {brier_score_loss(y_te, raw_xgb_te):.4f}")
print(f"Brier score iso (менше - краще): {brier_score_loss(y_te, calib_xgb_te):.4f}")


# --- ГРАФІК Reliability diagram ---
plt.figure(figsize=(7, 6))
for preds, name in [
    (raw_rf_te, "RF raw"),
    (calib_rf_te, "RF iso"),
    (raw_xgb_te, "XGB raw"),
    (calib_xgb_te, "XGB iso"),
]:
    frac_pos, mean_pred = calibration_curve(y_te, preds, n_bins=10)
    plt.plot(mean_pred, frac_pos, marker="o", label=name)
plt.plot([0, 1], [0, 1], "k--", label="Ideal")
plt.xlabel("Predicted probability")
plt.ylabel("Fraction of positives")
plt.legend()
plt.title("Reliability diagram (Калібрування ймовірностей)")
plt.tight_layout()
plt.show()

# --- OPTIMAL THRESHOLDING (P=R) ---
prec, rec, thresh = precision_recall_curve(y_te, calib_rf_te)
idx = np.argmin(np.abs(prec - rec))
best_thresh = thresh[idx]

print("\n--- Оптимальний поріг (Precision = Recall) ---")
print(f"Best thresh (RF): {best_thresh:.3f} → P/R: {prec[idx]:.3f}, {rec[idx]:.3f}")

rf_final = (calib_rf_te >= best_thresh).astype(int)
print("\n[Classification Report RF (Final, Tuned Threshold)]")
print(classification_report(y_te, rf_final, target_names=["H", "F"]))

# XGBoost
prec_x, rec_x, thresh_x = precision_recall_curve(y_te, calib_xgb_te)
idx_x = np.argmin(np.abs(prec_x - rec_x))
best_thresh_x = thresh_x[idx_x]
print(f"Best thresh (XGB): {best_thresh_x:.3f} → P/R: {prec_x[idx_x]:.3f}, {rec_x[idx_x]:.3f}")

xgb_final_pred = (calib_xgb_te >= best_thresh_x).astype(int)
print("\n[Classification Report XGB (Final, Tuned Threshold)]")
print(classification_report(y_te, xgb_final_pred, target_names=["H", "F"]))

# --- ГРАФІК Precision–Recall curve ---
plt.figure(figsize=(6, 4))
plt.plot(rec, prec, label="RF")
plt.plot(rec_x, prec_x, label="XGB")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision–Recall curve")
plt.legend()
plt.show()


# ==============================================================================
# КОМІРКА 19: ЗБЕРЕЖЕННЯ МОДЕЛЕЙ ТА КОРОЛЯЦІЙНІ МАТРИЦІ (Секція 17)
# ==============================================================================

print("\n--- Секція 19: Збереження моделей та Кореляційні матриці ---")

# Збереження фінальних моделей
with open(os.path.join(DATA_ROOT, "rf_rul_model.pkl"), "wb") as f:
    pickle.dump(rf, f)
print("Модель rf_rul_model.pkl збережена.")

with open(os.path.join(DATA_ROOT, "xgb_rul_model.pkl"), "wb") as f:
    pickle.dump(xgb_final, f)
print("Модель xgb_rul_model.pkl збережена.")


# --- ГРАФІК Кореляційна матриця ознак вібрації ---
vib_feats = [c for c in df_feats.columns if c not in ("file", "label")]
corr_vib = df_feats[vib_feats].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(corr_vib, cmap="viridis", center=0, cbar_kws={"shrink": 0.8})
plt.title("Correlation matrix — Vibration features")
plt.show()


# ==============================================================================
# КОМІРКА 20: ДОДАТКОВА ВІЗУАЛІЗАЦІЯ ФІЛЬТРАЦІЇ (Секція 18)
# ==============================================================================

print("\n--- Секція 20: Візуалізація фільтрації (Band-pass vs Wavelet) ---")

faulty_files = glob.glob(os.path.join(DATA_PATH_VIBRATION, "Faulty", "*.mat"))
sig_raw = load_signal_mat(faulty_files[0])[:, 0]

# 1. Band-pass фільтрація (старий метод)
sig_filt_bp = bandpass(sig_raw, low=5, high=100, fs=FS_VIB)

# 2. Wavelet деноїзинг (НОВИЙ метод)
try:
    sig_filt_w = wavelet_denoise(sig_raw, wavelet='db4', level=5, threshold_factor=0.5)
except Exception as e:
    sig_filt_w = sig_raw # Запобігання збою, якщо pywt не встановлений
    print(f"Помилка вейвлет-деноізингу: {e}. Використовується сирий сигнал.")


plt.figure(figsize=(12, 6))
# Обмежимо візуалізацію 5000 відліками для кращої наочності
plt.plot(sig_raw[:5000], color="gray", alpha=0.5, label="Raw X")
plt.plot(sig_filt_bp[:5000], color="orange", label="Filtered X (Band-pass)")
plt.plot(sig_filt_w[:5000], color="green", alpha=0.7, label="Filtered X (Wavelet Denoise)")
plt.title(f"Faulty {os.path.basename(faulty_files[0])}: Порівняння методів фільтрації")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.legend()
plt.tight_layout()
plt.show()
```
